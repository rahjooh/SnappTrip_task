## 4.5. 🔍 Diagnose & Fix Duplicate booking_id Issues

**Problem**: 5 duplicate `booking_id` values causing DBT tests to fail

**Failed Tests**:
- `unique_silver_booking_state_booking_id`
- `dbt_utils_unique_combination_of_columns_silver_booking_state_booking_id`

Let's diagnose and fix this data quality issue before proceeding to Gold layer.

````python
# 🔍 Diagnose Duplicate booking_id Issue
import os
from pyspark import SparkContext, SparkConf  
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count, desc, row_number
from pyspark.sql.window import Window

def create_diagnostic_spark():
    """Create Spark session for diagnostics"""
    try:
        if 'spark' in globals():
            globals()['spark'].stop()
        SparkContext._active_spark_context = None
        SparkSession._instantiatedSession = None
    except:
        pass
    
    os.environ['JAVA_HOME'] = '/opt/java/openjdk'
    
    return SparkSession.builder \
        .appName("DBT-Duplicate-Diagnostic") \
        .master("local[2]") \
        .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.4.3,org.apache.hadoop:hadoop-client:3.3.6") \
        .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
        .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
        .config("spark.sql.catalog.local.type", "hadoop") \
        .config("spark.sql.catalog.local.warehouse", "hdfs://namenode:9000/lakehouse") \
        .config("spark.driver.memory", "2g") \
        .getOrCreate()

# Create diagnostic session
diagnostic_spark = create_diagnostic_spark()
diagnostic_spark.sparkContext.setLogLevel("ERROR")
print("✅ Diagnostic Spark session created")
````

````python
# 📊 Analyze Silver Layer Duplicates
print("🔍 Analyzing silver_booking_state for duplicates...")

try:
    # Read the silver layer table
    silver_df = diagnostic_spark.table("local.silver.silver_booking_state")
    total_records = silver_df.count()
    unique_bookings = silver_df.select("booking_id").distinct().count()
    
    print(f"📋 Silver Layer Summary:")
    print(f"  Total records: {total_records}")
    print(f"  Unique booking_ids: {unique_bookings}")
    print(f"  Duplicate records: {total_records - unique_bookings}")
    
    # Find duplicates
    duplicates = silver_df.groupBy("booking_id").count().filter(col("count") > 1).orderBy(desc("count"))
    duplicate_count = duplicates.count()
    
    if duplicate_count > 0:
        print(f"\n❌ Found {duplicate_count} booking_ids with duplicates:")
        duplicates.show(10, False)
        
        # Show detailed duplicate records for first few
        duplicate_ids = [row['booking_id'] for row in duplicates.limit(3).collect()]
        for booking_id in duplicate_ids:
            print(f"\n🔍 Duplicate records for booking_id: {booking_id}")
            duplicate_records = silver_df.filter(col("booking_id") == booking_id).orderBy("updated_at", "created_at")
            duplicate_records.select("booking_id", "status", "price", "created_at", "updated_at").show(truncate=False)
    else:
        print("✅ No duplicates found - this is unexpected given the test failures")
        
except Exception as e:
    print(f"❌ Error reading silver layer: {e}")
    print("This might indicate the silver layer hasn't been created yet or there's a connection issue")
````

````python
# 💡 Generate Deduplication SQL Fix
print("\n💡 SQL Fix for silver_booking_state.sql model:")
print("="*60)

deduplication_sql = '''
{{ config(
    materialized='table',
    file_format='iceberg',
    location_root='hdfs://namenode:9000/lakehouse/silver/silver_booking_state'
) }}

WITH bronze_data AS (
    SELECT * FROM {{ ref('bronze_booking_state') }}
),

enriched AS (
    SELECT 
        b.booking_id,
        b.user_id,
        b.hotel_id,
        b.status,
        b.price,
        b.created_at,
        b.updated_at,
        h.city,
        h.star_rating,
        CURRENT_TIMESTAMP() as last_updated_ts
    FROM bronze_data b
    LEFT JOIN {{ ref('hotels') }} h ON b.hotel_id = h.hotel_id
),

-- 🔧 DEDUPLICATION: Keep latest record per booking_id
deduplicated AS (
    SELECT *,
        ROW_NUMBER() OVER (
            PARTITION BY booking_id 
            ORDER BY updated_at DESC, created_at DESC, last_updated_ts DESC
        ) as row_num
    FROM enriched
)

SELECT 
    booking_id,
    user_id,
    hotel_id,
    status,
    price,
    city,
    star_rating,
    created_at,
    updated_at,
    last_updated_ts
FROM deduplicated 
WHERE row_num = 1
'''

print(deduplication_sql)

print("\n🎯 Steps to Fix:")
print("1. Update dbt/models/silver/silver_booking_state.sql with the SQL above")
print("2. Run: dbt run --select silver_booking_state --target local")
print("3. Test: dbt test --select silver_booking_state --target local")
print("4. Verify uniqueness tests pass")
````

````python
# 🔧 Alternative: Direct DBT Run with Fix
print("\n🔧 Let's apply the fix directly...")

# First, let's check the current silver model
silver_model_path = Path('../dbt/models/silver/silver_booking_state.sql')

if silver_model_path.exists():
    print(f"📄 Current silver model found at: {silver_model_path}")
    
    # Read current content
    with open(silver_model_path, 'r') as f:
        current_content = f.read()
    
    print("\n📋 Current model content (first 500 chars):")
    print(current_content[:500] + "..." if len(current_content) > 500 else current_content)
    
    # Check if it already has deduplication logic
    has_dedup = "ROW_NUMBER" in current_content or "row_num" in current_content
    print(f"\n🔍 Deduplication logic present: {'✅ Yes' if has_dedup else '❌ No'}")
    
    if not has_dedup:
        print("\n⚠️ Model needs deduplication logic added")
        print("Would you like to:")
        print("  A) Backup current model and apply fix")
        print("  B) Just run tests to confirm current status")
        print("  C) Show current test results")
    else:
        print("\n✅ Model appears to have deduplication - issue might be elsewhere")
else:
    print(f"❌ Silver model not found at: {silver_model_path}")
    print("Creating the directory structure and model file...")
````

````python
# 🧪 Quick Test: Run DBT Test on Silver Layer Only
print("\n🧪 Testing silver layer uniqueness constraints...")

# Run just the failing tests
test_cmd = [
    'dbt', 'test', 
    '--select', 'silver_booking_state', 
    '--target', 'local',
    '--profiles-dir', './profiles'
]

print(f"📜 Running: {' '.join(test_cmd)}")
result = subprocess.run(test_cmd, capture_output=True, text=True, cwd='../dbt')

print("\n📊 Test Results:")
print(result.stdout)

if result.returncode == 0:
    print("\n✅ All silver layer tests passed! The issue is resolved.")
else:
    print(f"\n❌ Tests still failing. Return code: {result.returncode}")
    if result.stderr:
        print("\n🚨 Errors:")
        print(result.stderr)
    
    # Check specifically for the uniqueness test failures
    if "unique_silver_booking_state_booking_id" in result.stdout:
        print("\n🎯 The uniqueness test is still failing - deduplication fix needed")
    
# Clean up diagnostic session
try:
    diagnostic_spark.stop()
    print("\n📴 Diagnostic session stopped")
except:
    pass
````

## 8. View Lineage Graph (Optional)

````python
# Parse manifest.json to show lineage
manifest_path = DBT_PROJECT_DIR / 'target' / 'manifest.json'

if manifest_path.exists():
    with open(manifest_path) as f:
        manifest = json.load(f)
    
    print("📊 Data Lineage:")
    print("=" * 70)
    
    # Show model dependencies
    for node_id, node in manifest['nodes'].items():
        if node['resource_type'] == 'model':
            model_name = node['name']
            depends_on = node.get('depends_on', {}).get('nodes', [])
            
            print(f"\n📦 {model_name}")
            print(f"   Layer: {node.get('tags', [])}")
            print(f"   Depends on:")
            for dep in depends_on:
                dep_name = dep.split('.')[-1]
                print(f"     ↳ {dep_name}")
    
    print("\n" + "=" * 70)
    print("\n✅ Lineage extracted from manifest!")
else:
    print("⚠️  Manifest not found. Run 'dbt docs generate' first.")
````
