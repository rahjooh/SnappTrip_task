# Kafka Service - Producer & Consumer

This notebook demonstrates Kafka producer and consumer operations.

```python
import sys
sys.path.append('../')

from src.kafka.producer import KafkaProducer, produce_bookings
from src.kafka.consumer import KafkaConsumer, consume_topic
import pandas as pd
```

## 1. Load Sample Data

```python
# Load bookings data
bookings_df = pd.read_csv('../data/bookings_raw.csv')
print(f"Loaded {len(bookings_df)} bookings")
bookings_df.head()
```

## 2. Produce Messages to Kafka

```python
# Initialize producer
producer = KafkaProducer(bootstrap_servers='localhost:19092')

# Produce first 5 bookings
for _, row in bookings_df.head(5).iterrows():
    message = {
        'booking_id': str(row['booking_id']),
        'user_id': str(row['user_id']),
        'hotel_id': str(row['hotel_id']),
        'status': str(row['status']),
        'price': float(row['price']),
        'created_at': str(row['created_at']),
        'updated_at': str(row['updated_at'])
    }
    producer.produce('bookings_raw', key=message['booking_id'], value=message)
    print(f"Produced: {message['booking_id']}")

producer.flush()
print("✅ All messages produced!")
```

## 3. Consume Messages from Kafka

```python
# Initialize consumer
consumer = KafkaConsumer(bootstrap_servers='localhost:19092', group_id='notebook-consumer')

# Subscribe to topic
consumer.subscribe(['bookings_raw'])

# Consume messages
messages = []

def collect_message(msg):
    messages.append(msg)
    print(f"Consumed: {msg['key']} - {msg['value']['status']}")

# Consume 5 messages
consumer.consume(collect_message, max_messages=5)

print(f"\n✅ Consumed {len(messages)} messages")
```

## 4. Analyze Consumed Data

```python
# Convert to DataFrame
consumed_df = pd.DataFrame([msg['value'] for msg in messages])
consumed_df
```
