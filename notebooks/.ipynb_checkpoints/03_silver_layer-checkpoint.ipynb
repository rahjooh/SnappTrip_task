This notebook runs DBT to transform Bronze → Silver layer:
- Reads from: `file:///tmp/lakehouse/bronze/bookings_raw` (local Iceberg tables)
- Writes to: `file:///tmp/lakehouse/silver/silver_booking_state` (local Iceberg tables)
- Format: Iceberg (Snappy Parquet)
- Includes: Deduplication, enrichment, validation, data quality tests
- **Note**: Uses local filesystem for development. For production, use HDFS via DBT profiles.

```python
import sys
import os
import subprocess
sys.path.append('../')

# Set Java 17 for PySpark 3.5.0 only on Mac (Docker image has its own JAVA_HOME)
_java_home = '/Library/Java/JavaVirtualMachines/temurin-17.jdk/Contents/Home'
if os.path.exists(_java_home):
    os.environ['JAVA_HOME'] = _java_home

from pyspark.sql import SparkSession
```

## 1. Initialize Spark Session for DBT

```python
# Stop any existing Spark session
try:
    spark.stop()
    print("Stopped existing Spark session")
except:
    pass

# Create Spark session with local filesystem (for notebook development)
spark = SparkSession.builder \
    .appName("Silver-Layer-DBT") \
    .master("local[*]") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.4.3,org.apache.hadoop:hadoop-client:3.3.6") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.local.type", "hadoop") \
    .config("spark.sql.catalog.local.warehouse", "file:///tmp/lakehouse") \
    .config("spark.sql.iceberg.compression-codec", "snappy") \
    .config("spark.sql.parquet.compression.codec", "snappy") \
    .config("spark.sql.iceberg.write.format.default", "parquet") \
    .config("spark.driver.memory", os.environ.get("SPARK_DRIVER_MEMORY", "2g")) \
    .config("spark.executor.memory", os.environ.get("SPARK_EXECUTOR_MEMORY", "2g")) \
    .getOrCreate()

print("✅ Spark session initialized for DBT!")
print(f"   Version: {spark.version}")
print(f"   Warehouse: file:///tmp/lakehouse (local filesystem)")
print(f"   Compression: Snappy Parquet")
print(f"   Note: Using local filesystem for notebook development")
```

## 2. Verify Bronze Layer Data

```python
try:
    print("📋 Bronze Layer Tables:")
    spark.sql("SHOW TABLES IN local.bronze").show()

    print("\n📊 Bronze Bookings:")
    bronze_bookings = spark.table("local.bronze.bookings_raw")
    print(f"Count: {bronze_bookings.count()}")
    bronze_bookings.show(5)

    print("\n📊 Reference Hotels:")
    hotels = spark.table("local.reference.hotels")
    print(f"Count: {hotels.count()}")
    hotels.show()
except Exception as e:
    err = str(e)
    if "NoSuchNamespaceException" in err or "Namespace does not exist" in err:
        print("❌ Bronze namespace not found in catalog 'local' (warehouse: file:///tmp/lakehouse).")
        print("\n   To fix:")
        print("   1. Run 02_bronze_layer.ipynb first with warehouse file:///tmp/lakehouse")
        print("   2. Or copy bronze/reference data from HDFS into /tmp/lakehouse/ on this host")
        print("\n   Then re-run this cell.")
    else:
        raise
```

## 3. Run DBT Silver Layer Transformation

```python
# Repo root (notebooks/ is one level down)
_repo_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
os.chdir(os.path.join(_repo_root, 'dbt'))
os.environ["LAKEHOUSE_WAREHOUSE"] = "file:///tmp/lakehouse"

print("🚀 Running DBT Silver Layer transformation...\n")
print("=" * 70)

# When running dbt locally, use same warehouse as Spark so silver table is visible in this notebook
try:
    _wh = spark.conf.get("spark.sql.catalog.local.warehouse", "file:///tmp/lakehouse")
    os.environ["LAKEHOUSE_WAREHOUSE"] = _wh
except Exception:
    os.environ.setdefault("LAKEHOUSE_WAREHOUSE", "file:///tmp/lakehouse")

# Run DBT inside Docker (same network as HDFS) to avoid "RPC response has invalid length" on host
_docker_dir = os.path.join(_repo_root, 'docker')
if os.path.isfile(os.path.join(_docker_dir, 'docker-compose.yml')):
    result = subprocess.run(
        ['docker', 'compose', 'run', '--rm', 'dbt-run', 'run', '--select', 'tag:silver', '--profiles-dir', '.', '--profile', 'snapptrip', '--target', 'local_docker'],
        capture_output=True,
        text=True,
        cwd=_docker_dir
    )
    print("(DBT ran via Docker: target=local_docker, warehouse=namenode:9000)\n")
else:
    try:
        result = subprocess.run(
            ['dbt', 'run', '--select', 'tag:silver', '--profiles-dir', '.', '--profile', 'snapptrip', '--target', 'local'],
            capture_output=True,
            text=True
        )
        print("(DBT ran locally: target=local)\n")
    except FileNotFoundError:
        print("❌ 'dbt' CLI not found. Rebuild the notebook image: docker compose -f docker/docker-compose.yml build notebook")
        result = type('Result', (), {'returncode': 1, 'stdout': '', 'stderr': ''})()

print(result.stdout)
if result.stderr:
    print("\nWarnings/Errors:")
    print(result.stderr)

print("=" * 70)

if result.returncode == 0:
    print("\n✅ DBT Silver layer transformation completed successfully!")
else:
    print(f"\n❌ DBT run failed with exit code {result.returncode}")

os.chdir(os.path.join(_repo_root, 'notebooks'))
```

## 4. Run DBT Data Quality Tests

```python
os.chdir('../dbt')

print("🧪 Running DBT data quality tests for Silver layer...\n")
print("=" * 70)

# Run DBT tests
result = subprocess.run(
    ['dbt', 'test', '--select', 'tag:silver', '--profiles-dir', '.', '--profile', 'snapptrip', '--target', 'local'],
    capture_output=True,
    text=True
)

print(result.stdout)
if result.stderr:
    print("\nWarnings/Errors:")
    print(result.stderr)

print("=" * 70)

if result.returncode == 0:
    print("\n✅ All data quality tests passed!")
else:
    print(f"\n⚠️  Some tests failed (exit code {result.returncode})")
    print("   If unique_silver_booking_state_booking_id failed (duplicate booking_id): run")
    print("   dbt run --full-refresh --select silver_booking_state then re-run this cell.")

os.chdir('../notebooks')
```

## 5. Verify Silver Layer in HDFS

```python
print("📋 Silver Layer Tables:")
try:
    spark.sql("SHOW TABLES IN local.silver").show()
    print("\n📊 Silver Booking State:")
    silver_df = spark.table("local.silver.silver_booking_state")
    print(f"Count: {silver_df.count()}")
    print("\nSchema:")
    silver_df.printSchema()
    print("\nSample data:")
    silver_df.show(10, truncate=False)
except Exception as e:
    err = str(e)
    if "NoSuchNamespaceException" in err or "Namespace does not exist" in err or "silver" in err:
        _wh = "file:///tmp/lakehouse"
        try:
            _wh = spark.conf.get("spark.sql.catalog.local.warehouse", _wh)
        except Exception:
            pass
        print("❌ Silver namespace not found in catalog 'local' (warehouse:", _wh + ").")
        print("\n   To fix:")
        print("   1. Run the DBT Silver cell above (with Bronze data in the same warehouse).")
        print("   2. If DBT runs in Docker, Silver is on HDFS; use the same warehouse in this notebook or run DBT locally.")
        print("\n   Then re-run this cell.")
    else:
        raise
```

## 6. Check HDFS Storage for Silver Layer

```python
print("📁 HDFS Silver Layer Directory:")
print("\n" + "=" * 70)

result = subprocess.run(
    ['docker', 'exec', 'namenode', 'hdfs', 'dfs', '-ls', '-R', '/lakehouse/silver'],
    capture_output=True,
    text=True
)

print(result.stdout)

print("\n" + "=" * 70)
print("\n✅ Silver layer data persisted in HDFS!")
print("\n📍 HDFS Path:")
print("   • hdfs://namenode:9000/lakehouse/silver/silver_booking_state")
print("\n🌐 View in HDFS UI: http://localhost:9870/explorer.html#/lakehouse/silver")
```

## 7. Analyze Silver Layer Data

```python
# Status distribution
print("📊 Booking Status Distribution:")
silver_df.groupBy("status").count().orderBy("count", ascending=False).show()

# City distribution
print("\n📊 City Distribution:")
silver_df.groupBy("city").count().orderBy("count", ascending=False).show()

# Star rating distribution
print("\n📊 Star Rating Distribution:")
silver_df.groupBy("star_rating").count().orderBy("star_rating").show()

# Price statistics
print("\n📊 Price Statistics:")
silver_df.select(
    "price"
).describe().show()
```
